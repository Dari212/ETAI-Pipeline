# Prepocessing steps needed:
1. Missing values like '?', '-'
2. Different names for the same category 
3. Numerical values like strings
4. Negative values in numerical columns
5. Range of numerical values
6. Duplicate values
7. id column

```python
def standardize_missing_values(df, missing_values_types = ['?', '-'] ):
    for col in df.select_dtypes(include=["object", "category", "str"]).columns:
        df[col] = df[col].replace(missing_values_types, np.nan)
    return df
```

```python
df = standardize_missing_values(df)
```

```python
for col in df.select_dtypes(include=["object", "category", "str"]).columns:
    print(col, df[col].unique())
```

```python
def transform_cat_to_num(df, cols = ["prior_offenses", "priors_count"]):
    for col in cols:
        df[col] = pd.to_numeric(df[col], errors='coerce')
        # errors='coerce' turn non-numeric strings into NaN 
    return df
```

```python
df = transform_cat_to_num(df)
```

```python
# `Male`, `Female`
# `African-American`, `Caucasian`, `Hispanic`, `Asian`, `Native American`, `Other`
# `Low`, `Medium`, `High`
# `F`, `M`
```

```python
def change_categorical_col(df, cat_cols = ["score_text", "sex", "c_charge_degree", "race",]):
    #['Low', 'High', 'Medium', 'low', 'medium', 'LOW', 'MEDIUM', 'HIGH', 'high']
    map_score = {'low': 'Low', 'medium': 'Medium', 'high': 'High'}
    #['Male', 'male', 'female', 'Female', 'MALE', ' Male', 'Female ', nan, 'FEMALE']
    map_sex = {'male': 'Male', 'female': 'Female'}
    #['F', 'M', 'm', 'Misdemeanor', nan, 'Felony', 'f']
    map_charge_degree= {'m': 'M', 'f': 'F', 'misdemeanor': 'M', 'felony': 'F'} 
    #[ 'African-American','Caucasian', nan, 'Hispanic','Other',  'african-american',' African-American', 'HISPANIC','caucasian',
    # 'hispanic', 'CAUCASIAN','Native America', Caucasian',  'African American',  'AFRICAN-AMERICAN', 'Asian']
    map_race = {'african-american': 'African-American', 'caucasian': 'Caucasian', 'hispanic': 'Hispanic',
                'asian': 'Asian', 'other': 'Other', 'native american': 'Native American', 'african american': 'African-American' }
    df["score_text"] = df["score_text"].str.lower().str.strip().map(map_score)
    df["sex"] = df["sex"].str.lower().str.strip().map(map_sex)
    df["c_charge_degree"] = df["c_charge_degree"].str.lower().str.strip().map(map_charge_degree)
    df["race"] = df["race"].str.lower().str.strip().map(map_race)
    return df    
```

```python
df1 = change_categorical_col(df)
```

```python
def clean_numerical_cols(df, num_cols = ['juvenile_total', 'juv_fel_count', 'age_in_months', 'age']):
    df.loc[df['age'] < 0, 'age'] = np.nan
    df.loc[df['age_in_months']<0, 'age_in_months']= np.nan
    df.loc[df['juv_fel_count']<0, 'juv_fel_count']= np.nan
    df.loc[df['juvenile_total']<0, 'juvenile_total']= np.nan
    return df
```
